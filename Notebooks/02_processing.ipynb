## Creating new columns to answer the KPI questions
#### business questions:
1. How is the business doing overall?
   KPIs: Total Sales, Total Profit, Profit Margin %, Total Orders,
   Total Customers, Average Order Value

2. Is the business growing?
   KPIs: Sales and Profit by year, Year-over-Year growth %

3. Are there seasonal patterns?
   KPI: Sales by month (busy and slow periods)

4. Which categories and sub-categories bring the most sales and profit?
   KPIs: Sales, Profit, and Profit Margin % by Category and Sub-Category

5. Which sub-categories lose money?
   KPI: Sub-categories with negative total profit

6. Do discounts hurt profit?
   KPI: Profit Margin % by discount level (None, Standard, Medium, High)

7. Which regions and states perform best and worst?
   KPIs: Sales, Profit, and Profit Margin % by Region and State

8. Which customer segments are the most valuable, and how fast do we deliver?
   KPIs: Sales, Profit, and Average Order Value by Segment;
   Average Delivery Days by Ship Mode

```python
df['Year'] = df['Order Date'].dt.year  # for question number 2
df['Month'] = df['Order Date'].dt.month   # for question number 3
df['Delivery Days'] = (df['Ship Date'] - df['Order Date']).dt.days
df.isna().sum()
```

## Prepare Stage: New Columns

The questions need information that is not directly in the data, so I created these columns from existing ones. The cleaned data was loaded from the Parquet file, which keeps the column types (dates stay dates and Postal Code stays text). The CSV version does not store types, so anyone reading it must specify the dates and Postal Code when loading.

| New column | How it is built | Question it serves | Check | Result |
|---|---|---|---|---|
| Year | Year taken from Order Date | Q2: Is the business growing? | Unique values should be exactly 2014, 2015, 2016, 2017 | Only 2014, 2015, 2016 and 2017 present. Passed. |
| Month | Month number (1 to 12) taken from Order Date | Q3: Are there seasonal patterns? | Unique values should be exactly 1 to 12 | Values 1 to 12 only. Passed. |
| Delivery Days | Ship Date minus Order Date, converted to a whole number of days | Q8: How fast do we deliver? | Minimum should be 0 or higher, maximum should be a small number, and Same Day should average close to 0 | Minimum 0, maximum 7 days. Averages: Same Day 0.044 , First Class 2.183 , Second Class 3.238 , Standard Class 5.007. Passed. |
| Discount Level | Discount grouped into four levels: None (0%), Standard (above 0% up to 20%), Medium (above 20% up to 40%), High (above 40%) | Q6: Do discounts hurt profit? | Every row has a level (no empty values), and the counts per level add up to 9,993 | No empty values. Counts: None 4,798, Standard 3,803, Medium 459, High 933. Passed. |

### Notes on the columns

- Month is only the month number, without the year. It adds up all Januaries across the years, which is what the seasonality question needs. A monthly timeline will be built in Power BI from Order Date.
- Subtracting two dates gives a time difference, not a number, so `.dt.days` is used to get plain integers.
- Discount is stored as a fraction (0.2 means 20%).

### Discount Level: why these boundaries

- None is its own level because it is the baseline, with 48% of the rows (4,798).
- Standard covers 10%, 15% and 20% discounts. Most of its rows are exactly 20% (3,657 of 3,803), which is the usual promotion in this store, so I named it Standard and not Low.
- The boundaries at 20% and 40% were chosen after looking at the value counts of Discount, and before looking at profit, so the groups are not shaped to fit the answer.
- A value equal to a boundary goes into the lower group (0.20 is Standard, 0.40 is Medium).
- Expected row counts: None 4,798, Standard 3,803, Medium 459, High 933. The smallest level has 459 rows, which is enough for a reliable margin.

### Checks

- No missing values in the new columns:0 missing values in all four new columns.
- Final shape:9,993 rows and 24 columns.

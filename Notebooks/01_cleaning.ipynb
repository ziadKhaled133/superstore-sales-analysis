```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../Raw_data/Superstore.csv", encoding="latin1")
df
```

### reviewing the data information

```python
print(df.info())
print(20*" * ")
df.describe()
```

### checking nulls and duplicates


```python
print(df.isna().sum())
print()
print(df.duplicated().sum())
```

```python
# transforming the date columns to datetype

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df['Postal Code'] = df['Postal Code'].astype(str).str.zfill(5)
# i did this to make sure that all postal codes are five characters long, i also compensted for the missing leading zeros which removed when the column was integer format
```

```python
#dropping the row id since its useless

df.drop( columns=['Row ID'], inplace=True)
df
```

```python
#rounding the sales and profit columns to 2 decimal places

df['Sales'] = df['Sales'].round(2)
df['Profit'] = df['Profit'].round(2)
df
```

```python
# checking duplicated rows after droping the Row ID column
print(df.duplicated().sum())
```

```python
duplicated_rows = df[df.duplicated(keep=False)]
duplicated_rows
```

```python
#dropping that row since its the same order with identical values so maybe its a mistake while entering the data

df.drop_duplicates(keep='first', inplace=True)
print(df.duplicated().sum())
```

```python
#checking the unique values in each column to catch up wrong spelling or duplicated 

display(df['Ship Mode'].unique())
display(df['Segment'].unique())
display(df['Country'].unique())
display(df['City'].unique())
display(df['State'].unique())
display(df['Region'].unique())
display(df['Category'].unique())
display(df['Sub-Category'].unique())
```
